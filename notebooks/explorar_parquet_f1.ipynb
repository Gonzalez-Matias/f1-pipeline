# Exploración de datos consolidados F1

Este notebook abre los archivos CSV consolidados de F1 ubicados en `airflow-f1/include/output/`.

```python
import pandas as pd
import os

base_path = 'airflow-f1/include/output'
full_path = os.path.join(base_path, 'f1_all_full.csv')
results_path = os.path.join(base_path, 'f1_all_results.csv')
```

## 1. f1_all_full.csv

```python
df_full = pd.read_csv(full_path, sep=';')
print('Shape:', df_full.shape)
print('Columnas:', df_full.columns.tolist())
df_full.head()
```

```python
df_full.info()
```

```python
df_full.describe(include='all').T
```

## 2. f1_all_results.csv

```python
df_results = pd.read_csv(results_path, sep=';')
print('Shape:', df_results.shape)
print('Columnas:', df_results.columns.tolist())
df_results.head()
```

```python
df_results.info()
```

```python
df_results.describe(include='all').T
```

## 3. Cruce rápido (opcional)

```python
# Ejemplo: ver años distintos presentes en cada dataset
if 'Year' in df_full.columns:
    print('Años en full:', sorted(df_full['Year'].dropna().unique()))
if 'Year' in df_results.columns:
    print('Años en results:', sorted(df_results['Year'].dropna().unique()))
```
